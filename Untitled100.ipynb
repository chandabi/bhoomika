```python
import os
os.system("pip install -q gradio")

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import gradio as gr

# Dataset folder: one sub-folder per disease (e.g. eczema/, melanoma/ ...)
DATA_DIR = "/content/skin_dataset"
IMG_SIZE = (128, 128)

# ---- Helper to download or generate a small sample dataset if empty ----
if not os.path.exists(DATA_DIR) or len(os.listdir(DATA_DIR)) == 0:
    print("Dataset folder is empty or missing. Setting up a small sample structure...")
    os.makedirs(os.path.join(DATA_DIR, "eczema"), exist_ok=True)
    os.makedirs(os.path.join(DATA_DIR, "melanoma"), exist_ok=True)
    os.makedirs(os.path.join(DATA_DIR, "acne"), exist_ok=True)
    os.makedirs(os.path.join(DATA_DIR, "psoriasis"), exist_ok=True)
    os.makedirs(os.path.join(DATA_DIR, "seborrheic_keratosis"), exist_ok=True)
    os.makedirs(os.path.join(DATA_DIR, "squamous_cell_carcinoma"), exist_ok=True)
    os.makedirs(os.path.join(DATA_DIR, "vascular_lesion"), exist_ok=True)

    # Generate 10 dummy images per class so the dataset has data to load and split
    for i in range(10):
        img_data = np.random.randint(0, 255, (128, 128, 3), dtype=np.uint8)
        tf.keras.utils.save_img(os.path.join(DATA_DIR, "eczema", f"eczema_{i}.jpg"), img_data)
        tf.keras.utils.save_img(os.path.join(DATA_DIR, "melanoma", f"melanoma_{i}.jpg"), img_data)
        tf.keras.utils.save_img(os.path.join(DATA_DIR, "acne", f"acne_{i}.jpg"), img_data)
        tf.keras.utils.save_img(os.path.join(DATA_DIR, "psoriasis", f"psoriasis_{i}.jpg"), img_data)
        tf.keras.utils.save_img(os.path.join(DATA_DIR, "seborrheic_keratosis", f"seborrheic_keratosis_{i}.jpg"), img_data)
        tf.keras.utils.save_img(os.path.join(DATA_DIR, "squamous_cell_carcinoma", f"squamous_cell_carcinoma_{i}.jpg"), img_data)
        tf.keras.utils.save_img(os.path.join(DATA_DIR, "vascular_lesion", f"vascular_lesion_{i}.jpg"), img_data)
# ---- 1. Load the images ----
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR, validation_split=0.2, subset="training",
    seed=42, image_size=IMG_SIZE, batch_size=32)
val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR, validation_split=0.2, subset="validation",
    seed=42, image_size=IMG_SIZE, batch_size=32)
class_names = train_ds.class_names
print("Diseases:", class_names)

# ---- 2. Build the CNN ----
model = models.Sequential([
    layers.Input(shape=IMG_SIZE + (3,)),
    layers.Rescaling(1.0 / 255),                       # pixels 0-255 -> 0-1
    layers.Conv2D(32, 3, activation="relu"), layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation="relu"), layers.MaxPooling2D(),
    layers.Conv2D(128, 3, activation="relu"), layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dropout(0.5),
    layers.Dense(128, activation="relu"),
    layers.Dense(len(class_names), activation="softmax"),
])
model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

# ---- 3. Train ----
model.fit(train_ds, validation_data=val_ds, epochs=10)

# ---- 4. Prediction function for the app ----
def predict(img):
    img = img.convert("RGB").resize(IMG_SIZE)
    x = np.expand_dims(np.array(img, dtype="float32"), axis=0)
    probs = model.predict(x, verbose=0)[0]
    return {class_names[i]: float(probs[i]) for i in range(len(class_names))}

# ---- 5. Launch the app ----
gr.Interface(
    fn=predict,
    inputs=gr.Image(type="pil", label="Upload skin image"),
    outputs=gr.Label(num_top_classes=3),
    title="Skin Disease Detection",
    description="Educational project only - not a medical diagnosis.",
).launch(share=True)
```
