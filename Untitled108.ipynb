```python
import gradio as gr
import random
import time

PICTURES = [
    "🍎", "🐱", "🐶", "🍓",
    "🦋", "🌻", "🚗", "🐼",
    "🍉", "🦁", "🐸", "🍇",
    "🐰", "🌈", "🐵", "⚽"
]

def create_game(level="Easy"):
    count = {"Easy": 4, "Medium": 6, "Hard": 8}[level]

    selected = random.sample(PICTURES, count)
    cards = selected * 2
    random.shuffle(cards)

    return {
        "cards": cards,
        "opened": [],
        "matched": [],
        "score": 0,
        "moves": 0,
        "start": time.time(),
        "elapsed": 0,
        "finished": False,
        "level": level
    }

def get_outputs(s, message=""):
    result = []

    for i in range(16):
        if i >= len(s["cards"]):
            result.append(gr.update(value="", visible=False))
        elif i in s["matched"] or i in s["opened"]:
            result.append(gr.update(value=s["cards"][i], visible=True))
        else:
            result.append(gr.update(value="❓", visible=True))

    seconds = s["elapsed"] if s["finished"] else time.time() - s["start"]

    status = message or (
        "🏆 YOU WON! 🎉🥳🎊" if s["finished"]
        else "🧠 Find the matching pictures!"
    )

    celebration = ""

    if s["finished"]:
        celebration = """
        <div class="celebration">
            <div class="trophy">🏆</div>
            <h1>CONGRATULATIONS!</h1>
            <h2>🎉 🥳 🎊</h2>
            <h2>YOU WON THE GAME!</h2>
            <p>Excellent Memory! ❤️</p>
        </div>
        """

    return (
        [s] + result +
        [
            status,
            f"⭐ Score: {s['score']}",
            f"🎯 Moves: {s['moves']}",
            f"⏱️ Time: {int(seconds)//60:02d}:{int(seconds)%60:02d}",
            celebration
        ]
    )

def new_game(level):
    return get_outputs(create_game(level), "🎮 New Game Started!")

def reset_game(s):
    if s is None:
        return new_game("Easy")

    level = s["level"]
    cards = s["cards"].copy()
    random.shuffle(cards)

    new_state = {
        "cards": cards,
        "opened": [],
        "matched": [],
        "score": 0,
        "moves": 0,
        "start": time.time(),
        "elapsed": 0,
        "finished": False,
        "level": level
    }

    return get_outputs(new_state, "🔄 Game Reset!")

def play(index, s):
    if s is None or s["finished"]:
        return get_outputs(s)

    if index in s["matched"]:
        return get_outputs(s)

    if len(s["opened"]) == 2:
        s["opened"] = []

    if index in s["opened"]:
        return get_outputs(s)

    s["opened"].append(index)

    if len(s["opened"]) == 2:
        a, b = s["opened"]
        s["moves"] += 1

        if s["cards"][a] == s["cards"][b]:
            s["matched"].extend([a, b])
            s["opened"] = []
            s["score"] += 10

            if len(s["matched"]) == len(s["cards"]):
                s["finished"] = True
                s["elapsed"] = time.time() - s["start"]

                return get_outputs(s, "🏆 YOU WON! 🎉🥳")

            return get_outputs(s, "✅ Correct Match! +10 Points")

    return get_outputs(s)

def ai_solve(s):
    if s is None or s["finished"]:
        return get_outputs(s)

    s["matched"] = list(range(len(s["cards"])))
    s["opened"] = []
    s["score"] = len(s["cards"]) // 2 * 10
    s["finished"] = True
    s["elapsed"] = time.time() - s["start"]

    return get_outputs(s, "🤖 AI SOLVED THE GAME! 🏆")

def hint(s):
    if s is None or s["finished"]:
        return get_outputs(s)

    remaining = [
        i for i in range(len(s["cards"]))
        if i not in s["matched"]
    ]

    if len(remaining) >= 2:
        first = remaining[0]
        second = next(
            i for i in remaining[1:]
            if s["cards"][i] == s["cards"][first]
        )

        s["opened"] = [first, second]
        return get_outputs(s, "💡 Hint: Look at these two cards!")

    return get_outputs(s)

def clock(s):
    if s is None:
        return "⏱️ Time: 00:00"

    seconds = s["elapsed"] if s["finished"] else time.time()-s["start"]
    return f"⏱️ Time: {int(seconds)//60:02d}:{int(seconds)%60:02d}"


CSS = """
.gradio-container {
    background: linear-gradient(-45deg,#171044,#532b91,#087c91,#25104c);
    background-size: 400% 400%;
    animation: bg 12s ease infinite;
    color: white !important;
}
@keyframes bg {
    0% {background-position:0% 50%}
    50% {background-position:100% 50%}
    100% {background-position:0% 50%}
}
h1,h2,h3,p,label {
    color: white !important;
}
.memory button {
    background: linear-gradient(135deg,#ff69b4,#7556ff) !important;
    color: white !important;
    border: 3px solid white !important;
    border-radius: 18px !important;
    min-height: 85px !important;
    font-size: 35px !important;
    box-shadow: 0 5px 15px #0005;
}
.celebration {
    background: linear-gradient(135deg,#ff9a44,#ff4d9d);
    text-align: center;
    padding: 25px;
    border-radius: 25px;
    animation: celebrate 1s infinite alternate;
}
.trophy {
    font-size: 85px;
    animation: bounce 1s infinite;
}
@keyframes bounce {
    50% {transform: translateY(-20px) rotate(10deg)}
}
@keyframes celebrate {
    to {transform: scale(1.02)}
}
"""

with gr.Blocks(
    css=CSS,
    theme=gr.themes.Soft(),
    title="AI Image Memory Game"
) as demo:

    gr.Markdown("""
    # 🧠 IMAGE MEMORY MATCHING GAME
    ### 🎮 Test Your Memory and Win Trophies!
    """)

    state = gr.State(create_game())

    level = gr.Dropdown(
        ["Easy", "Medium", "Hard"],
        value="Easy",
        label="Select Difficulty"
    )

    with gr.Row():
        score = gr.Markdown("⭐ Score: 0")
        moves = gr.Markdown("🎯 Moves: 0")
        timer = gr.Markdown("⏱️ Time: 00:00")

    status = gr.Markdown("🧠 Find the matching pictures!")

    buttons = []

    for row in range(4):
        with gr.Row():
            for col in range(4):
                i = row * 4 + col
                btn = gr.Button(
                    "❓",
                    elem_classes=["memory"]
                )
                buttons.append(btn)

    celebration = gr.HTML("")

    with gr.Row():
        new_btn = gr.Button("🆕 New Game", variant="primary")
        reset_btn = gr.Button("🔄 Reset")

    with gr.Row():
        hint_btn = gr.Button("💡 Hint")
        ai_btn = gr.Button("🤖 AI Solve", variant="primary")

    outputs = [
        state, *buttons, status,
        score, moves, timer, celebration
    ]

    for i, btn in enumerate(buttons):
        btn.click(
            fn=lambda s, idx=i: play(idx, s),
            inputs=[state],
            outputs=outputs
        )

    new_btn.click(
        new_game,
        inputs=[level],
        outputs=outputs
    )

    reset_btn.click(
        reset_game,
        inputs=[state],
        outputs=outputs
    )

    hint_btn.click(
        hint,
        inputs=[state],
        outputs=outputs
    )

    ai_btn.click(
        ai_solve,
        inputs=[state],
        outputs=outputs
    )

    gr.Timer(1).tick(
        clock,
        inputs=[state],
        outputs=[timer]
    )

    gr.Markdown("""
    ### 🏆 Match all the pictures and become a champion!
    """)

demo.launch(share=True)
```
