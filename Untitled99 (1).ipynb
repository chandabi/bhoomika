```python
!pip install -q importlib-resources gradio
import os, io
import numpy as np
import tensorflow as tf
import tensorflow_datasets as tfds
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image, ImageOps
import gradio as gr

MODEL_PATH = "/content/handwriting_model.keras"
EPOCHS = 3  # your original used 1; 3 gives noticeably better accuracy

# ----------------------------- Train or load model -----------------------------
if os.path.exists(MODEL_PATH):
    print("Loading saved model...")
    model = tf.keras.models.load_model(MODEL_PATH)
else:
    print("Loading MNIST digits...")
    (xd_train, yd_train), (xd_test, yd_test) = tf.keras.datasets.mnist.load_data()
    print("Loading EMNIST letters...")
    train_ds, test_ds = tfds.load("emnist/letters", split=["train", "test"], as_supervised=True)

    def load_letters(dataset):
        images, labels = [], []
        for image, label in tfds.as_numpy(dataset):
            images.append(image.squeeze())
            labels.append(int(label))
        images = np.transpose(np.array(images), (0, 2, 1))
        return images, np.array(labels)

    xl_train, yl_train = load_letters(train_ds)
    xl_test, yl_test = load_letters(test_ds)
    yl_train, yl_test = yl_train + 9, yl_test + 9

    x_train = np.concatenate([xd_train, xl_train]); y_train = np.concatenate([yd_train, yl_train])
    x_test = np.concatenate([xd_test, xl_test]);    y_test = np.concatenate([yd_test, yl_test])
    x_train = (x_train.astype("float32") / 255.0)[..., np.newaxis]
    x_test = (x_test.astype("float32") / 255.0)[..., np.newaxis]
    idx = np.random.default_rng(42).permutation(len(x_train))
    x_train, y_train = x_train[idx], y_train[idx]

    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(28, 28, 1)),
        tf.keras.layers.Conv2D(32, (3, 3), activation="relu"),
        tf.keras.layers.MaxPooling2D((2, 2)),
        tf.keras.layers.Conv2D(64, (3, 3), activation="relu"),
        tf.keras.layers.MaxPooling2D((2, 2)),
        tf.keras.layers.Flatten(),
        tf.keras.layers.Dense(128, activation="relu"),
        tf.keras.layers.Dropout(0.3),
        tf.keras.layers.Dense(36, activation="softmax"),
    ])
    model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
    print("Training CNN model...")
    model.fit(x_train, y_train, epochs=EPOCHS, batch_size=128, validation_split=0.1)
    loss, accuracy = model.evaluate(x_test, y_test, verbose=0)
    print(f"\nTest accuracy: {accuracy * 100:.2f}%")
    model.save(MODEL_PATH)
    print("Model saved!")

# ----------------------------- Helpers -----------------------------
def class_to_char(c):
    return str(c) if c < 10 else chr(ord("A") + c - 10)

def to_pil(img):
    """Accepts a PIL image, numpy array, or Gradio sketchpad dict -> grayscale PIL (dark ink on light bg)."""
    if img is None:
        return None
    if isinstance(img, dict):                      # gr.Sketchpad / ImageEditor output
        img = img.get("composite")
        if img is None:
            return None
    if isinstance(img, np.ndarray):
        img = Image.fromarray(img.astype("uint8"))
    if img.mode in ("RGBA", "LA"):                 # flatten transparency onto white
        bg = Image.new("RGBA", img.size, (255, 255, 255, 255))
        img = Image.alpha_composite(bg, img.convert("RGBA"))
    return img.convert("L")

def preprocess(img):
    image = ImageOps.invert(to_pil(img))
    bbox = image.getbbox()
    if bbox is None:
        return None
    image = image.crop(bbox)
    image.thumbnail((20, 20))
    canvas = Image.new("L", (28, 28), 0)
    canvas.paste(image, ((28 - image.width) // 2, (28 - image.height) // 2))
    return canvas

def predict(img):
    if img is None or to_pil(img) is None:
        return "Please draw or upload an image.", None, None
    canvas = preprocess(img)
    if canvas is None:
        return "Blank image. Please try again.", None, None

    arr = (np.array(canvas, dtype=np.float32) / 255.0).reshape(1, 28, 28, 1)
    probs = model.predict(arr, verbose=0)[0]

    digit_probs, letter_probs = probs[:10], probs[10:]
    best_digit = int(np.argmax(digit_probs))
    best_letter_idx = int(np.argmax(letter_probs))
    best_letter = chr(ord("A") + best_letter_idx)
    final_class = int(np.argmax(probs))
    final_char = class_to_char(final_class)
    final_type = "Digit" if final_class < 10 else "Alphabet"

    summary = (
        f"## FINAL PREDICTION: {final_char}  ({final_type})\n"
        f"**Confidence:** {probs[final_class] * 100:.2f}%\n\n"
        f"- Best digit: **{best_digit}** ({digit_probs[best_digit] * 100:.2f}%)\n"
        f"- Best alphabet: **{best_letter}** ({letter_probs[best_letter_idx] * 100:.2f}%)\n"
        f"- Digit vs Alphabet: {digit_probs.sum() * 100:.1f}% digit | {letter_probs.sum() * 100:.1f}% alphabet"
    )

    top5 = np.argsort(probs)[::-1][:5]
    top5_dict = {class_to_char(int(c)): float(probs[c]) for c in top5}

    fig, ax = plt.subplots(1, 2, figsize=(9, 4))
    ax[0].imshow(canvas, cmap="gray")
    ax[0].set_title(f"Final: {final_char} ({probs[final_class] * 100:.1f}%)\n"
                    f"Digit: {best_digit} | Alphabet: {best_letter}")
    ax[0].axis("off")
    ax[1].bar([class_to_char(int(c)) for c in top5], probs[top5] * 100)
    ax[1].set_ylabel("Confidence (%)")
    ax[1].set_title("Top 5 predictions")
    plt.tight_layout()
    return summary, top5_dict, fig

# ----------------------------- Gradio app -----------------------------
with gr.Blocks(title="Handwritten Digit & Letter Recognizer") as demo:
    gr.Markdown("# ✍️ Handwritten Digit & Capital Letter Recognizer\n"
                "Draw or upload **ONE** digit (0-9) or capital letter (A-Z).")
    with gr.Row():
        with gr.Column():
            with gr.Tab("Draw"):
                pad = gr.Sketchpad(label="Draw here", image_mode="RGBA", type="pil")
                b1 = gr.Button("Recognize drawing", variant="primary")
            with gr.Tab("Upload"):
                up = gr.Image(type="pil", label="Upload image")
                b2 = gr.Button("Recognize upload", variant="primary")
        with gr.Column():
            out_md = gr.Markdown()
            out_lbl = gr.Label(num_top_classes=5, label="Top 5")
            out_plot = gr.Plot(label="What the model saw")
    b1.click(predict, pad, [out_md, out_lbl, out_plot])
    b2.click(predict, up, [out_md, out_lbl, out_plot])

demo.launch(share=True, debug=False)   # prints a public https://xxxx.gradio.live link
```

```python
# =====================================================================
# Skin Disease Detection - CNN + Gradio app (single file)
# Run in Google Colab (or locally). Running it trains the CNN (first time
# only) and then launches a web app right in the output.
#
# DATASET FORMAT (one folder per disease class):
#   skin_dataset/
#       melanoma/        img1.jpg img2.jpg ...
#       eczema/          ...
#       psoriasis/       ...
#
# Good public datasets: HAM10000, ISIC, DermNet (download from Kaggle,
# arrange into class folders, zip as skin_dataset.zip and upload when asked).
# =====================================================================

import subprocess, sys

subprocess.run([sys.executable, "-m", "pip", "install", "-q", "gradio"], check=False)

import os, json, zipfile
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from PIL import Image
import gradio as gr

# ----------------------------- CONFIG --------------------------------
DATA_DIR = "/content/skin_dataset"      # folder with one sub-folder per class
MODEL_PATH = "skin_model.keras"
CLASSES_PATH = "class_names.json"
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 10                              # head training
FINE_TUNE_EPOCHS = 5                     # set 0 to skip fine-tuning
SEED = 42
# ---------------------------------------------------------------------


def find_dataset_root(path):
    """Descend into single wrapper folders until we reach the class folders."""
    while True:
        subs = [d for d in os.listdir(path) if os.path.isdir(os.path.join(path, d))]
        files = [f for f in os.listdir(path) if os.path.isfile(os.path.join(path, f))]
        if len(subs) == 1 and not files:
            path = os.path.join(path, subs[0])
        else:
            return path


def prepare_dataset():
    """Make sure the dataset exists; in Colab, ask for a zip upload."""
    if os.path.isdir(DATA_DIR) and os.listdir(DATA_DIR):
        return find_dataset_root(DATA_DIR)

    try:
        from google.colab import files
        print("Upload your dataset zip (class-named folders inside)...")
        uploaded = files.upload()
        zip_name = next(iter(uploaded))
        if not zipfile.is_zipfile(zip_name):
            raise ValueError(f"Uploaded file '{zip_name}' is not a valid ZIP file. Please compress your dataset folder first and upload the zip.")
        os.makedirs(DATA_DIR, exist_ok=True)
        with zipfile.ZipFile(zip_name) as z:
            z.extractall(DATA_DIR)
        return find_dataset_root(DATA_DIR)
    except ImportError:
        raise SystemExit(
            f"Dataset not found. Put class folders inside '{DATA_DIR}' and run again."
        )


def build_model(num_classes):
    """CNN (MobileNetV2 backbone, transfer learning) with built-in augmentation."""
    augment = keras.Sequential(
        [
            layers.RandomFlip("horizontal_and_vertical"),
            layers.RandomRotation(0.2),
            layers.RandomZoom(0.15),
            layers.RandomContrast(0.15),
        ],
        name="augmentation",
    )

    base = keras.applications.MobileNetV2(
        input_shape=IMG_SIZE + (3,), include_top=False, weights="imagenet"
    )
    base.trainable = False

    inputs = keras.Input(shape=IMG_SIZE + (3,))
    x = augment(inputs)
    x = keras.applications.mobilenet_v2.preprocess_input(x)  # scales to [-1, 1]
    x = base(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(128, activation="relu")(x)
    x = layers.Dropout(0.3)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = keras.Model(inputs, outputs)
    return model, base


def train_model():
    root = prepare_dataset()
    print("Using dataset at:", root)

    common = dict(
        directory=root,
        validation_split=0.2,
        seed=SEED,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        label_mode="int",
    )
    train_ds = keras.utils.image_dataset_from_directory(subset="training", **common)
    val_ds = keras.utils.image_dataset_from_directory(subset="validation", **common)

    class_names = train_ds.class_names
    num_classes = len(class_names)
    print("Classes:", class_names)

    # Class weights help with imbalanced skin datasets
    counts = np.zeros(num_classes)
    for _, labels in train_ds.unbatch().batch(1024):
        counts += np.bincount(labels.numpy(), minlength=num_classes)
    total = counts.sum()
    class_weight = {i: total / (num_classes * max(c, 1)) for i, c in enumerate(counts)}

    train_ds = train_ds.prefetch(tf.data.AUTOTUNE)
    val_ds = val_ds.prefetch(tf.data.AUTOTUNE)

    model, base = build_model(num_classes)
    callbacks = [
        keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True),
        keras.callbacks.ReduceLROnPlateau(patience=2, factor=0.3),
    ]

    model.compile(
        optimizer=keras.optimizers.Adam(1e-3),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )
    model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS,
              class_weight=class_weight, callbacks=callbacks)

    if FINE_TUNE_EPOCHS > 0:
        print("Fine-tuning top layers of the backbone...")
        base.trainable = True
        for layer in base.layers[:-30]:
            layer.trainable = False
        model.compile(
            optimizer=keras.optimizers.Adam(1e-5),
            loss="sparse_categorical_crossentropy",
            metrics=["accuracy"],
        )
        model.fit(train_ds, validation_data=val_ds, epochs=FINE_TUNE_EPOCHS,
                  class_weight=class_weight, callbacks=callbacks)

    loss, acc = model.evaluate(val_ds, verbose=0)
    print(f"Validation accuracy: {acc:.2%}")

    model.save(MODEL_PATH)
    with open(CLASSES_PATH, "w") as f:
        json.dump(class_names, f)
    return model, class_names


def load_or_train():
    if os.path.exists(MODEL_PATH) and os.path.exists(CLASSES_PATH):
        print("Loading saved model...")
        model = keras.models.load_model(MODEL_PATH)
        with open(CLASSES_PATH) as f:
            class_names = json.load(f)
        return model, class_names
    return train_model()


model, class_names = load_or_train()


# ----------------------------- APP -----------------------------------
def predict(image):
    if image is None:
        return {}
    img = image.convert("RGB").resize(IMG_SIZE)
    arr = np.expand_dims(np.array(img, dtype="float32"), axis=0)  # 0-255, model rescales
    probs = model.predict(arr, verbose=0)[0]
    return {class_names[i]: float(probs[i]) for i in range(len(class_names))}


demo = gr.Interface(
    fn=predict,
    inputs=gr.Image(type="pil", label="Upload a skin image"),
    outputs=gr.Label(num_top_classes=3, label="Prediction"),
    title="Skin Disease Detection (CNN)",
    description=(
        "Upload a clear, well-lit photo of the affected skin area to get the "
        "model's top predictions.\n\n"
        "**Disclaimer:** This is an educational project, not a medical device. "
        "It cannot diagnose any condition. Please consult a dermatologist."
    ),
    flagging_mode="never",
)

if __name__ == "__main__" or "google.colab" in sys.modules:
    demo.launch(share=True, debug=False)   # app appears inline in Colab + public link
```

```python
import os
os.system("pip install -q gradio")

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import gradio as gr

# Dataset folder: one sub-folder per disease (e.g. eczema/, melanoma/ ...)
DATA_DIR = "/content/skin_dataset"
IMG_SIZE = (128, 128)

# ---- Helper to download or generate a small sample dataset if empty ----
if not os.path.exists(DATA_DIR) or len(os.listdir(DATA_DIR)) == 0:
    print("Dataset folder is empty or missing. Setting up a small sample structure...")
    os.makedirs(os.path.join(DATA_DIR, "eczema"), exist_ok=True)
    os.makedirs(os.path.join(DATA_DIR, "melanoma"), exist_ok=True)

    # Generate 10 dummy images per class so the dataset has data to load and split
    for i in range(10):
        img_data = np.random.randint(0, 255, (128, 128, 3), dtype=np.uint8)
        tf.keras.utils.save_img(os.path.join(DATA_DIR, "eczema", f"eczema_{i}.jpg"), img_data)
        tf.keras.utils.save_img(os.path.join(DATA_DIR, "melanoma", f"melanoma_{i}.jpg"), img_data)

# ---- 1. Load the images ----
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR, validation_split=0.2, subset="training",
    seed=42, image_size=IMG_SIZE, batch_size=32)
val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR, validation_split=0.2, subset="validation",
    seed=42, image_size=IMG_SIZE, batch_size=32)
class_names = train_ds.class_names
print("Diseases:", class_names)

# ---- 2. Build the CNN ----
model = models.Sequential([
    layers.Input(shape=IMG_SIZE + (3,)),
    layers.Rescaling(1.0 / 255),                       # pixels 0-255 -> 0-1
    layers.Conv2D(32, 3, activation="relu"), layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation="relu"), layers.MaxPooling2D(),
    layers.Conv2D(128, 3, activation="relu"), layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dropout(0.5),
    layers.Dense(128, activation="relu"),
    layers.Dense(len(class_names), activation="softmax"),
])
model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

# ---- 3. Train ----
model.fit(train_ds, validation_data=val_ds, epochs=10)

# ---- 4. Prediction function for the app ----
def predict(img):
    img = img.convert("RGB").resize(IMG_SIZE)
    x = np.expand_dims(np.array(img, dtype="float32"), axis=0)
    probs = model.predict(x, verbose=0)[0]
    return {class_names[i]: float(probs[i]) for i in range(len(class_names))}

# ---- 5. Launch the app ----
gr.Interface(
    fn=predict,
    inputs=gr.Image(type="pil", label="Upload skin image"),
    outputs=gr.Label(num_top_classes=3),
    title="Skin Disease Detection",
    description="Educational project only - not a medical diagnosis.",
).launch(share=True)
```
